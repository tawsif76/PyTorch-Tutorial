```python
import torch
```

```python
x = torch.tensor(3.0, requires_grad = True)
y = x ** 2

y
```

```python
y.backward()
x.grad #dy/dx
```

```python
x = torch.tensor(3.0, requires_grad = True)
y = x ** 2
z = torch.sin(y)

x, y, z
```

```python
z.backward()

x.grad, y.grad
```

x.grad.zero_() to clear prev grad values, otherwise it keeps accumulating.
